```python
import pandas as pd

teclo_data = pd.read_csv('data/Telco_Customer_Churn.csv')
teclo_df = teclo_data.copy()

#teclo_df.head()
#teclo_df.info()
teclo_df.describe()
```

2. Чистка данных

- заменить пропуски:
 - числовые → медиана
 - категориальные → "Unknown"

```python
print(teclo_df.isnull().sum())
```

3. Фильтрация

найти клиентов:
- у которых MonthlyCharges > 70
- и Contract == "Month-to-month"

👉 цель: логические условия

```python
teclo_df_fitred = teclo_df[teclo_df['MonthlyCharges'] > 70][teclo_df['Contract'] == 'Month-to-month']

display(teclo_df_fitred)
```

4. Feature Engineering

создай новые колонки:

- total_spent = tenure * MonthlyCharges
- is_senior = 1 если SeniorCitizen == 1 иначе 0

👉 цель: работа с колонками

```python
teclo_df['total_spent'] = teclo_df['tenure'] * teclo_df['MonthlyCharges']

for col in teclo_df['SeniorCitizen']:
   if col == 1:
      teclo_df['is_senior'] = 1
   else:
      teclo_df['is_senior'] = 0

teclo_df['is_senior']
```
