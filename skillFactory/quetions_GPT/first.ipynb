## Задача 3 — агрегация (medium)
Сделай через groupby():
- Среднюю цену дома по районам (Suburb)
- Отсортируй по убыванию
- Выведи топ-5 самых дорогих районов

groupby — это способ сгруппировать данные по какому-то признаку, чтобы потом что-то с ними сделать.

Агрегация (aggregation)

Агрегация — это сжатие данных внутри группы до одного значения.

Типа:
- среднее (mean)
- сумма (sum)
- количество (count)
- максимум (max)
- минимум (min)

```python
melb_df_arg = melb_df.groupby('Suburb')['Price'].mean()
melb_df_arg = melb_df_arg.sort_values(ascending=False)
top_5_suburbs = melb_df_arg.head(5)
display(top_5_suburbs) 
```

ascending — это направление сортировки
- ascending=True → по возрастанию (от меньшего к большему)
- ascending=False → по убыванию (от большего к меньшему)

## Задача 5 — мини Data Science (hard)

Подготовь данные для ML:
- Удали строки с пропусками

Выбери признаки:
- Rooms
- Distance
- Landsize

Целевая переменная: Price



any() проверяет: есть ли хотя бы одно True значение

```python
melb_df_copy = melb_df.copy()

def display_null_values(df):
    for column in df.columns:
        # any() проверяет: есть ли хотя бы одно True значение
        if df[column].isnull().any():       
            # Проверяем, есть ли пропущенные значения в столбце
            if pd.api.types.is_numeric_dtype(df[column]):
            # Если столбец числовой (int или float), заменяем пропуски средним значением
                df[column].fillna(df[column].mean(), inplace=True)
            else:
            # Если столбец категориальный (object или string), заменяем пропуски модой (самым частым значением)
                df[column].fillna(df[column].mode()[0], inplace=True)
        return df  # Возвращаем измененный датафрейм

melb_df_copy = display_null_values(melb_df_copy)

melb_df_copy = melb_df_copy[['Rooms', 'Distance', 'Landsize']]

display(melb_df_copy.isnull().sum())
```
