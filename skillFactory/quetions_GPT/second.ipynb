## 1. Базовый groupby

Посчитай средний spend по city

```python
import pandas as pd

df = pd.DataFrame({
    'user': ['A','B','C','D','E','F','G','H'],
    'age': [23, 35, 23, 41, 35, 29, 41, 23],
    'city': ['Moscow','SPB','Moscow','Kazan','SPB','Kazan','Moscow','SPB'],
    'category': ['A','B','A','C','B','A','C','B'],
    'spend': [100, 200, 150, 300, 250, 120, 400, 180]
})

df.groupby('city')['spend'].mean()

```

## 2. Несколько агрегатов

По city найди:

- средний spend
- максимальный spend
- количество пользователей

```python
mask1 = df.groupby('city')['spend'].mean()
mask2 = df.groupby('city')['spend'].max()
mask3 = df.groupby('spend')['user'].count()

print(round(mask1))
print()
print(round(mask2))
print()
print(round(mask3))    
```

## 3. groupby + сортировка

Отсортируй города по убыванию среднего spend

```python
df['Средний spend'] = round(df.groupby('city')['spend'].transform('mean'))
df.sort_values(by='Средний spend')
```
