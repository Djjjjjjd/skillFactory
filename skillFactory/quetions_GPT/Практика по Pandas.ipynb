````python
import pandas as pd
import numpy as np

data = {
    'employee_id': range(1, 41),

    'name': [
        'Alex', 'John', 'Anna', 'Mike', 'Sara',
        'Tom', 'Kate', 'Bob', 'Lisa', 'David'
    ] * 4,

    'department': [
        'IT', 'HR', 'Finance', 'IT', 'Marketing',
        'Finance', 'HR', 'IT', 'Marketing', 'Finance'
    ] * 4,

    'city': [
        'Moscow', 'SPB', 'Kazan', 'Moscow', 'SPB',
        'Kazan', 'Moscow', 'SPB', 'Kazan', 'Moscow'
    ] * 4,

    'salary': [
        120000, 80000, 95000, 110000, 70000,
        99000, 85000, 130000, 72000, 105000
    ] * 4,

    'age': [
        25, 32, 29, 41, 24,
        36, 28, 45, 27, 38
    ] * 4,

    'experience_years': [
        1, 5, 3, 10, 1,
        7, 2, 15, 2, 8
    ] * 4,

    'bonus_percent': [
        10, 15, np.nan, 20, 5,
        12, 8, 25, np.nan, 18
    ] * 4,

    'performance_score': [
        'A', 'B', 'A', 'A', 'C',
        'B', 'B', 'A', 'C', 'B'
    ] * 4,

    'employment_date': pd.date_range(
        start='2021-01-10',
        periods=40,
        freq='MS'
    ).astype(str),

    'remote': [
        'Yes', 'No', 'Yes', 'Yes', 'No',
        'No', 'Yes', 'No', 'Yes', 'No'
    ] * 4
}

df = pd.DataFrame(data)

# Добавим пропуски
df.loc[5, 'salary'] = np.nan
df.loc[12, 'city'] = np.nan
df.loc[20, 'age'] = np.nan

# Сохраняем CSV
df.to_csv('employees.csv', index=False)

print(df.head())
````

Задание 5
Условие

Создай столбец:
- salary_level

Правила:
```
< 90000 → Low
90000–110000 → Middle

110000 → High

Использовать
apply()
```

Ожидаемый результат

Категории зарплат.

Подсказка

Сделай отдельную функцию.

````python
def slary_level_function(salary):
    if salary < 90000:
        return 'Low'
    elif 90000 <= salary < 110000:
        return 'Medium'
    else:
        return 'High'
    
em_df['salary_level'] = em_df['salary'].apply(slary_level_function)
em_df.head()
````
