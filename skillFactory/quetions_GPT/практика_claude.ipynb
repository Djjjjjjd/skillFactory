- Создай таблицу: строки = Regionname, столбцы = Type (h/u/t),
  значения = средняя цена. Используй pivot_table()
- В каком регионе самая большая разница в цене между h и u?

```python
sorted_df = melb_df.pivot_table(index='Regionname',columns='Type', aggfunc={'Price':'mean'})
sorted_df['diff_h_u'] = sorted_df['Price']['h'] - sorted_df['Price']['u']
sorted_df = sorted_df.sort_values(by='diff_h_u', ascending=False)
display(sorted_df.head(10))
```

- Заполни пропуски в CouncilArea значением 'Unknown'
- Заполни пропуски в Bedroom, Bathroom, Car медианой по каждому столбцу
- После заполнения проверь: df.isnull().sum() должен вернуть 0

```python
# замена пропущенных значений в колонке CouncilArea на 'Unknown'
melb_df['CouncilArea'] = melb_df['CouncilArea'].fillna('Unknown')
# замена пропущенных значений в колонке Bedroom, Bathroom, Car на медианное значение
melb_df['Bedroom'] = melb_df['Bedroom'].fillna(melb_df['Bedroom'].median())
melb_df['Bathroom'] = melb_df['Bathroom'].fillna(melb_df['Bathroom'].median())
melb_df['Car'] = melb_df['Car'].fillna(melb_df['Car'].median())

melb_df.isnull().sum()
```

- Переведи колонку Date в datetime
- Создай новые колонки: Year (год продажи), Month (месяц)
- Найди: в каком месяце продавалось больше всего объектов?

```python
# преобразование колонки Date в формат datetime и создание новых колонок Year и Month
melb_df['Date'] = pd.to_datetime(melb_df['Date'], dayfirst=True)
melb_df['Year'] = melb_df['Date'].dt.year
melb_df['Month'] = melb_df['Date'].dt.month
# средняя цена в зависимости от месяца продажи
avg_price_month = melb_df.groupby('Month')['Price'].mean()
display(avg_price_month)
```

- Создай колонку PricePerRoom = Price / Rooms
- Создай колонку AgeAtSale = год продажи - YearBuilt
- Создай бинарный флаг IsExpensive: 1 если цена > медианной цены, иначе 0

```python
# средняя цена за комнату
melb_df['PricePerRoom'] = melb_df['Price'] / melb_df['Rooms']
# возраст дома на момент продажи
melb_df['AgeAtSale'] = melb_df['Year'] - melb_df['YearBuilt']
# маркировка дорогих домов
melb_df['IsExpensive'] = melb_df.apply(lambda x: 1 if x['Price'] > melb_df['Price'].median() else 0, axis=1)
```
